```python
import numpy as np 

def mask_3(n, s):
    mask_3 = np.triu(np.ones((n,n)), -s) - np.triu(np.ones((n,n)),s+1)
    return mask_3


mask_3 = mask_3(9,1)
mask_1 = mask_3[:,::-1]
print(mask_1,'\n', mask_3)


```
